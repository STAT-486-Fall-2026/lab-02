## Section 1: Introduction to Scikit-learn

[Scikit-learn](https://scikit-learn.org/stable/)  is one of the most widely used Python packages for machine learning. It provides efficient implementations of many common machine learning algorithms and offers a clean, consistent, and easy-to-use API.

### Estimators in Scikit-learn
The core concept in scikit-learn is the estimator. An estimator is any object that learns from data by calling the `.fit()` method. There are two main types of estimators in scikit-learn:
1. Transformers
2. Predictors
---




#### 1. Transformers

A transformer is an estimator that performs a **transformation** on data. Examples of transformations include **standardizing features** (subtracting the mean and dividing by the standard deviation) or **filling missing values** in a dataset.

Transformers follow a two-step process:
1.	`.fit()` – Performs computations based on the training data.
2.	`.transform()` – Applies the transformation to the data and returns the transformed dataset.

**Important:** The `.fit()` method updates the transformer object but does not return any output. The .transform() method returns the transformed data.

For convenience, scikit-learn also provides a combined method:
* `.fit_transform()` – This method performs both steps at once: fitting the transformer and applying the transformation.

✅ Example of a Transformer:
```{python}
from sklearn.preprocessing import StandardScaler

# Create a transformer object
scaler = StandardScaler()

# Fit the transformer on the training data
scaler.fit(X_train)

# Transform the training data
X_train_scaled = scaler.transform(X_train)

# Alternatively, do both in one step
X_train_scaled = scaler.fit_transform(X_train)
```
---

#### 2. Predictors
A **predictor** is an estimator that can make predictions. For example, a Linear Regression model or a Random Forest Classifier are predictors.

Predictors follow a similar two-step process:
1.	`.fit()` – Trains the model using the training data.
2.	`.predict()` – Makes predictions on new data.

*Important:* Just like transformers, the .fit() method updates the predictor object but does not return any output. The .predict() method returns the model’s predictions.

For classification models, predictors may also have a:
* `.predict_proba()` – Returns the predicted class probabilities instead of class labels.

✅ Example of a Predictor:
```
from sklearn.linear_model import LinearRegression

# Create a predictor object
model = LinearRegression()

# Fit the model on the training data
model.fit(X_train_scaled, y_train)

# Make predictions on the test data
y_pred = model.predict(X_test_scaled)
```
---



### 📝 Why Does This Matter?

The **Estimator API** is a core feature of scikit-learn. Once you understand how to use the `.fit()`, `.transform()`, and `.predict()` methods, you can easily switch between different machine learning models and techniques with minimal changes to your code.

✅ Summary of Key Methods:
|Method|Transformer|Predictor|
|------|-----------|---------|
|`.fit()`| Learns the transformation | Trains the model|
|`.transform()`|Applies the transformation | N/A|
|`.fit_transform()`|Fits and transforms in one step | N/A |
|`.predict()` | N/A | Makes predictions |
|`.predict_proba()` |N/A| Returns class probabilities (for classifiers)|
